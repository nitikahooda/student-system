```python
import pandas as pd
import numpy as np
```

```python
data=pd.read_csv("student.csv")
df=pd.DataFrame(data)
df
```

```python
df.columns
```

```python
df.dtypes
```

```python
df["total_marks"]=df[["Python","Maths","DBMS","English"]].sum(axis=1)
df
```

```python
df["percentage"]=(df["total_marks"])/4
df
```

```python
def grade(percentage):
    if percentage>=90:
        return  ("A+")
    elif percentage>=80:
        return  ("A")
    elif percentage>=70:
        return ("B")
    elif percentage>=60:
        return ("C")
    elif percentage>=50:
        return ("D")
    elif percentage<50:
        return ("F")
```

```python
df["grade"]=df["percentage"].apply(grade)
df
```

```python
df["Rank"]=df["percentage"].rank(ascending=False).astype(int)
df
```

```python
def status(percentage):
    if percentage>=90:
        return  ("Excellent")
    elif percentage>=80:
        return  ("Very good")
    elif percentage>=70:
        return ("Good")
    elif percentage>=60:
        return ("Pass")
    elif percentage>=50:
        return ("Pass")
    elif percentage<50:
        return ("Fail")

df["Status"]=df["percentage"].apply(status)
df
```
