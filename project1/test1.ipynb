```python
df["total_marks"]=df[["Python","Maths","DBMS","English"]].sum(axis=1)
df
```

```python
df["percentage"]=(df["total_marks"])/4
df
```

```python
def grade(percentage):
    if percentage>=90:
        return  ("A+")
    elif percentage>=80:
        return  ("A")
    elif percentage>=70:
        return ("B")
    elif percentage>=60:
        return ("C")
    elif percentage>=50:
        return ("D")
    elif percentage<50:
        return ("F")
```

```python
df["grade"]=df["percentage"].apply(grade)
df
```
